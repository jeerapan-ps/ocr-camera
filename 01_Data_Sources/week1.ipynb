```python
import pytchat
import pandas as pd
chat = pytchat.create(video_id="xW-hj937DkI")

dList = []
aList = []
mList = []

i = 0
while chat.is_alive():
  for c in chat.get().sync_items():
    print(f"{c.datetime} [{c.author.name}]- {c.message}")
    dList.append(c.datetime)
    aList.append(c.author.name)
    mList.append(c.message)
    i = i+1
    if i == 10:
      chat.terminate()
      break

# dictionary of lists
dict = {'dateTime': dList, 'authorName': aList, 'Message': mList}

df = pd.DataFrame(dict)
df.head()

from google.colab import files
df.to_csv("data.csv", encoding="utf_8_sig", index=False) #cp874, #TIS-620
files.download("data.csv")
```

```python
from playwright.async_api import async_playwright
import asyncio

async def scrape_data():
    async with async_playwright() as p:
        # Launch browser (set headless=False to watch visually)
        browser = await p.chromium.launch(headless=True)
        page = await browser.new_page()

        # Navigate to target page
        await page.goto("https://books.toscrape.com/")

        # Extract main header text
        title = await page.locator("h1").inner_text()
        print(f"Page Title: {title}")

        # Extract multiple items
        books = page.locator("article.product_pod")
        count = await books.count()

        for i in range(count):
            book = books.nth(i)
            name = await book.locator("h3 a").get_attribute("title")
            price = await book.locator(".price_color").inner_text()
            print(f"{name} - {price}")

        await browser.close()

# Run the async function
await scrape_data()
```

```python
from playwright.async_api import async_playwright
import asyncio

async def scrape_data():
    async with async_playwright() as p:
        # Launch browser (set headless=False to watch visually)
        browser = await p.chromium.launch(headless=True)
        page = await browser.new_page()

        # Navigate to target page
        await page.goto("https://www.goldtraders.or.th/")

        # Extract main header text
        title = await page.locator("h1").inner_text()
        print(f"Page Title: {title}")

        # Extract multiple items
        container = page.locator(".lg\\:grid-cols-4")

        await container.locator("span.text-green-600").first.wait_for(
            state="visible",
            timeout=15000
        )

        print(await page.locator('span[data-slot="select-value"]').inner_text())

        boxes = container.locator(
            ":scope > div:not(.order-last)"
        )

        for i in range(await boxes.count()):
            texts = await boxes.nth(i).locator("span").all_inner_texts()

            if len(texts) >= 2:
                if "ทอง" in texts[0]:
                    print("_" * 10)
                print(f"{texts[0]}: {texts[1].strip()}")

        await browser.close()

# Run the async function
await scrape_data()
```

```python
from playwright.async_api import async_playwright
import asyncio

async def scrape_data():
    async with async_playwright() as p:
        # Launch browser (set headless=False to watch visually)
        browser = await p.chromium.launch(headless=True)
        page = await browser.new_page()

        # Navigate to target page
        await page.goto("https://www.premierleague.com/en/stats/all-time")

        # Extract main header text
        title = await page.locator("h2").first.inner_text()
        print(f"Page Title: {title}")

        # Extract multiple items
        div = page.locator(
            'div[data-stats-type="players"][data-stats-metric="goals"]'
        )

        await div.locator(".stats-leaderboard__pos").first.wait_for(
            state="visible",
            timeout=15000
        )

        players = div.locator("li")
        count = await players.count()
        #print(count)

        for i in range(count):
            print("_" * 20)
            print(f"อันดับที่ {await players.nth(i).locator(".stats-leaderboard__pos").inner_text()}")
            print(f"ชื่อ : {await players.nth(i).locator(".stats-leaderboard__name").inner_text()}")
            print(f"จำนวนประตู : {await players.nth(i).locator(".stats-leaderboard__stat-value").inner_text()}")

        await browser.close()

# Run the async function
await scrape_data()
```
